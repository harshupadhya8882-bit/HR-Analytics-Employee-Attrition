```python
bins=[18,25,35,45,55,65]
labels=["18-25","26-35","36-45","46-55","56-65"]
df["Age_group"]=pd.cut(df["Age"],bins=bins,labels=labels,include_lowest=True)
```

```python
department_attrition = (
    pd.crosstab(
        df["Department"],
        df["Attrition"],
        normalize="index"
    ) * 100
).sort_values(by="Yes", ascending=False)
print(department_attrition)
```

```python
department_attrition.plot(
    kind="bar",
    figsize=(6,4),
    title="Attrition Rate by Department"
)

plt.ylabel("Attrition Rate (%)")
plt.xlabel("Department")
plt.xticks(rotation=0)
plt.show()
```

```python
age_attrition = (
    pd.crosstab(
        df["Age_group"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

age_attrition["Yes"].sort_values(ascending=False).plot(
    kind="bar",
    figsize=(6,4),
    title="Attrition Rate by Age Group",
    color="orange"
)

plt.ylabel("Attrition Rate (%)")
plt.xlabel("Age Group")
plt.xticks(rotation=0)
plt.show()
```

```python
jobrole_attrition = (
    pd.crosstab(
        df["JobRole"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

jobrole_attrition["Yes"].sort_values().plot(
    kind="barh",
    figsize=(8,5),
    title="Attrition Rate by Job Role",
    color="green"
)

plt.xlabel("Attrition Rate (%)")
plt.ylabel("Job Role")
plt.show()
```

```python
monthly_income_by_jobrole = (
    df.groupby("JobRole")["MonthlyIncome"]
      .mean()
      .sort_values(ascending=False)
)

monthly_income_by_jobrole.plot(
    kind="barh",
    figsize=(8,5),
    color="steelblue",
    title="Average Monthly Income by Job Role"
)

plt.xlabel("Average Monthly Income")
plt.ylabel("Job Role")
plt.show()
```

```python
overtime_attrition = (
    pd.crosstab(
        df["OverTime"],
        df["Attrition"],
        normalize="index"
    ) * 100
)

overtime_attrition["Yes"].plot(
    kind="bar",
    figsize=(5,4),
    color="red",
    title="Attrition Rate by Overtime"
)

plt.ylabel("Attrition Rate (%)")
plt.xlabel("OverTime")
plt.xticks(rotation=0)
plt.show()
```
